### Bar Chart

## Line Chart
Question: Does Jumia show the biggest discounts at the top the black friday page

```python
jumia_data ['listing_source'].unique()
```

```python
main_page = jumia_data[jumia_data['listing_source'] == 'Main Black Friday page']

# Matplotib
# Work out the discount of each page first

avg_page = (main_page.groupby("listing_page")["current_price_ksh"].mean())

# Create line chart

plt.figure.figuresize= (15,6)
plt.plot(avg_page.index, avg_page.values, marker = "o", linewidth = 1.8)
plt.title("Average product price by listing page")
plt.xlabel("Listing page")
plt.ylabel("Average price ksh")
plt.grid(True, alpha= 0.3)
plt.xticks(range(0, 60, 2))
plt.show()
```

```python
# seaborn 
plt.figure(figsize= (15, 6))
sns.lineplot(data= main_page, x="listing_page", y= "current_price_ksh", marker = "o", linewidth= 1.8)
plt.title("Average product price by listing page")
plt.xlabel("Listing page")
plt.ylabel("Average price Ksh")
plt.grid(True, alpha= 0.3)
plt.xticks(range(0, 60, 2))
plt.show()
```

## Pie chart
Question: What share of products come from offcial stores?

```python
share = jumia_data["is_official_store"].value_counts()
print(share)

explode = (0, 0.1)
plt.pie(share,labels= ["Non-offcial store", "Official store"], autopct = "%1.0f%%", explode = explode, startangle = 90)
plt.title("Store Distribution")
plt.show()
```

Question: Which discount band does most of the products fall? Donut chart

```python
discount = jumia_data["discount_band"].value_counts()
print(discount)

plt.pie(discount, labels = ["40-59%", "20-39%", "60%+", "0-19%"], autopct= "%1.0f%%", wedgeprops={"width": 0.4}, explode= (0.05, 0.05, 0.05, 0.05))
plt.title("Discount Band Distribution")
plt.savefig(r"../data/donut.jpeg", dpi= 300, transparent= True)
plt.show()
```
